```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import os
import sys

sys.path.append(os.path.abspath(".."))

from src.load_data import load_all_data
from src.get_dfs import get_dfs
dfs = load_all_data()

(circuits,
constructors,
constructor_standings,
drivers,
driver_standings,
qualifying,
races,
results) = get_dfs(dfs)

```

```python
# Find drivers with missing date of birth
missing_dob = drivers[drivers["dob"].isnull()]

# Display relevant columns
missing_dob[[
    "driver_id",
    "nationality"
]]
```

```python
# Create clean copy
circuits_clean = circuits.copy()

# Drop unnecessary columns
circuits_clean = circuits_clean.drop(
    columns=["lat", "lng", "url"]
)

# Verify result
circuits_clean.head()

circuits_clean.to_csv(
    "../data/processed/circuits_clean.csv",
    index=False
)
```

```python
# Create clean copy
constructors_clean = constructors.copy()

# Drop unnecessary columns
constructors_clean = constructors_clean.drop(
    columns=["url"]
)

# Verify result
constructors_clean.head()

constructors_clean.to_csv(
    "../data/processed/constructors_clean.csv",
    index=False
)
```

```python
# Create clean copy
constructor_standings_clean = constructor_standings.copy()

# Drop unnecessary columns
constructor_standings_clean = constructor_standings_clean.drop(
    columns=["constructor"]
)

# Save cleaned dataframe
constructor_standings_clean.to_csv(
    "../data/processed/constructor_standings_clean.csv",
    index=False
)

# Verify result
constructor_standings_clean.head()
```

```python
# Create clean copy
drivers_clean = drivers.copy()

# Drop unnecessary columns
drivers_clean = drivers_clean.drop(
    columns=["code", "number", "url"]
)

# Handle missing values

# Fill missing nationality values
drivers_clean["nationality"] = (
    drivers_clean["nationality"]
    .fillna("Unknown")
)

# Leave missing DOB values unchanged for now
# Historical driver records are incomplete,
# and DOB may still be useful later.

# Verify result
display(drivers_clean.head())

# Check remaining missing values
print(drivers_clean.isnull().sum())

# Save cleaned dataframe
drivers_clean.to_csv(
    "../data/processed/drivers_clean.csv",
    index=False
)
```

```python
# Create clean copy
driver_standings_clean = driver_standings.copy()

# No columns removed
# Missing values in 'position' are kept for now,
# since drivers with 0 points may not receive
# an official standing position.

# Verify result
display(driver_standings_clean.head())

# Check missing values
print(driver_standings_clean.isnull().sum())

# Save cleaned dataframe
driver_standings_clean.to_csv(
    "../data/processed/driver_standings_clean.csv",
    index=False
)
```

```python
qualifying_clean = qualifying.copy()

# No columns removed

# Missing values in q1, q2, and q3 are kept for now
# because they contain meaningful information:
# - DNF/no valid lap
# - elimination before later qualifying sessions
# - historical qualifying format differences

# Verify result
display(qualifying_clean.head())

# Check missing values
print(qualifying_clean.isnull().sum())

# Save cleaned dataframe
qualifying_clean.to_csv(
    "../data/processed/qualifying_clean.csv",
    index=False
)
```

```python
races_clean = races.copy()

# Drop unnecessary columns
races_clean = races_clean.drop(
    columns=["url"]
)

# Missing values in 'time' are kept for now
# because historical race start times were
# not consistently documented.

# Verify result
display(races_clean.head())

# Check missing values
print(races_clean.isnull().sum())

# Save cleaned dataframe
races_clean.to_csv(
    "../data/processed/races_clean.csv",
    index=False
)
```

```python
results_clean = results.copy()

# Drop unnecessary columns
results_clean = results_clean.drop(
    columns=["race_name", "constructor"]
)

# Missing values are kept for now because:
# - 'time' only exists for drivers finishing
#   within one lap of the winner
# - 'fastest_lap' and 'fastest_lap_rank'
#   were not consistently recorded historically

# Verify result
display(results_clean.head())

# Check missing values
print(results_clean.isnull().sum())

# Save cleaned dataframe
results_clean.to_csv(
    "../data/processed/results_clean.csv",
    index=False
)
```
