```python
from pathlib import Path
import json

root = Path(r"C:\Users\Ander\Desktop\F1-project\data\raw")

telemetry_files = list(root.rglob("*_tel.json"))

bad_files = []

for i, file in enumerate(telemetry_files):

    if i % 100 == 0:
        print(i)

    try:

        with open(file, "r", encoding="utf-8") as f:
            data = json.load(f)["tel"]

        for key, values in data.items():

            if any(v is None for v in values):

                bad_files.append({
                    "file": str(file),
                    "column": key
                })

    except Exception as e:

        bad_files.append({
            "file": str(file),
            "error": str(e)
        })

print(f"Bad files: {len(bad_files)}")
```

```python
import ast
import pandas as pd

weather_path = (
    r"C:\Users\Ander\Desktop\F1-project\data"
    r"\processed\Abu Dhabi Grand Prix\Race\weather.parquet"
)

weather = pd.read_parquet(weather_path)

# Convert stringified lists back to lists
cols = [
    "wT",
    "wAT",
    "wH",
    "wP",
    "wR",
    "wTT",
    "wWD",
    "wWS",
]

parsed = {}

for col in cols:

    parsed[col] = ast.literal_eval(
        weather.loc[0, col]
    )

# Build proper dataframe
weather_expanded = pd.DataFrame({
    "time": parsed["wT"],
    "air_temp": parsed["wAT"],
    "humidity": parsed["wH"],
    "pressure": parsed["wP"],
    "rain": parsed["wR"],
    "track_temp": parsed["wTT"],
    "wind_dir": parsed["wWD"],
    "wind_speed": parsed["wWS"],
})

display(weather_expanded.head())

print(weather_expanded.shape)
```

```python
from pathlib import Path
import pandas as pd

processed_root = Path(
    r"C:\Users\Ander\Desktop\F1-project\data\processed"
)

records = []

EXCLUDE_FILES = {
    "meta_race.parquet",
}

for file in processed_root.rglob("*.parquet"):

    if file.name in EXCLUDE_FILES:
        continue

    parts = file.parts

    filename = file.stem

    # -----------------------------
    # Parse metadata
    # -----------------------------

    try:
        race = parts[-3]
        session = parts[-2]
    except:
        race = None
        session = None

    # -----------------------------
    # Determine file type
    # -----------------------------

    if filename.endswith("_laps"):

        file_type = "laps"
        driver = filename.replace("_laps", "")

    elif filename in [
        "weather",
        "corners",
        "drivers",
        "rcm",
        "session_laptimes",
    ]:

        file_type = filename
        driver = None

    else:

        file_type = "telemetry"
        driver = filename

    records.append({
        "path": str(file),
        "race": race,
        "session": session,
        "type": file_type,
        "driver": driver,
    })

registry = pd.DataFrame(records)

display(registry.head())

print(registry.shape)
```

```python
from src.data.load_driver_telemetry import (
    load_driver_session,
    load_session_laptimes_file
)

import ast

low_sample_laps = []
dnfdrivers = []

for row in registry.itertuples():

    if row.type != "telemetry":
        continue

    if row.session != "Race":
        continue

    telemetry, _ = load_driver_session(
        registry,
        row.race,
        row.session,
        row.driver
    )

    session_laptimes = load_session_laptimes_file(
        registry,
        row.race,
        row.session
    )

    # official race lap count
    official_race_laps = max(
        ast.literal_eval(session_laptimes["lap"].iloc[0])
    )

    # driver completed laps
    driver_max_lap = telemetry["lap"].max()

    # probable DNF
    is_dnf = (official_race_laps - driver_max_lap) > 2

    # samples per lap
    samples_per_lap = (
        telemetry
        .sort_values("lap")
        .groupby("lap")
        .size()
    )

    # laps under threshold
    under_500 = samples_per_lap[samples_per_lap < 500]

    # store low-sample laps
    for lap, samples in under_500.items():

        low_sample_laps.append({
            "race": row.race,
            "driver": row.driver,
            "lap": lap,
            "samples": samples
        })

    # store DNF drivers
    if is_dnf:

        dnfdrivers.append({
            "race": row.race,
            "driver": row.driver,
            "driver_max_lap": int(driver_max_lap),
            "official_race_laps": int(official_race_laps)
        })

print("LOW SAMPLE LAPS")
for item in low_sample_laps:
    print(item)

print("\nDNF DRIVERS")
for item in dnfdrivers:
    print(item)
```
